### Criterio utilizado para columna 'clasificacion'

Sabemos que la clasificacion de desempeño mínima observada es de 55 y la máxima de 99. Además, los cuartiles son aproximadamente:
- Q1 ≈ **63**
- Q2 (mediana) ≈ **75**
- Q3 ≈ **88.25**

Para simplificar en clase, definiremos los niveles así:

- **Baja**: `calificacion_desempeno < 63`
- **Media**: `4.5 <= calificacion_desempeno <= 88`
- **Alta**: `calificacion_desempeno > 88`

```python
bins = [0, 63, 88, 99]
labels = ["Baja", "Media", "Alta"]

df_modified["clasificacion"] = pd.cut(
    df_modified["calificacion_desempeno"],
    bins=bins,
    labels=labels,
    include_lowest=True
)

df_modified[["colaborador","calificacion_desempeno", "clasificacion"]].head(10)
```
